```python
product_profit = df.groupby("Product Name")["Gross Profit"].sum().sort_values(ascending=False)

product_profit.head(10)
```

```python
plt.figure(figsize=(12,6))

product_profit.head(10).plot(kind="bar")

plt.title("Top 10 Most Profitable Products")
plt.xlabel("Product Name")
plt.ylabel("Gross Profit")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

- These products should be prioritized in inventory planning and promotions.
- Lower-ranked products should be reviewed for pricing or cost optimization.

```python
pareto_sales = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False)

pareto_sales
```

```python
pareto_df = pareto_sales.reset_index()

pareto_df.columns = ["Product Name", "Sales"]

pareto_df.head()
```

```python
pareto_df["Cumulative Sales"] = pareto_df["Sales"].cumsum()
```

```python
pareto_df["Cumulative %"] = (
    pareto_df["Cumulative Sales"] / pareto_df["Sales"].sum()
) * 100
```

```python
pareto_df.head(10)
```

```python
plt.figure(figsize=(12,6))

plt.bar(pareto_df["Product Name"], pareto_df["Sales"])

plt.plot(
    pareto_df["Product Name"],
    pareto_df["Cumulative %"],
    color="red",
    marker="o"
)

plt.xticks(rotation=90)

plt.title("Pareto Analysis - Revenue Contribution")

plt.ylabel("Sales / Cumulative %")

plt.tight_layout()

plt.show()
```

## Business Insight

- The Pareto chart helps identify whether a small number of products contribute the majority of revenue.
- Products appearing before the 80% cumulative point are strategically important.
- Heavy dependence on a few products may increase business risk if demand changes.

```python
plt.figure(figsize=(10,6))

plt.scatter(df["Cost"], df["Sales"])

plt.title("Cost vs Sales Analysis")
plt.xlabel("Cost")
plt.ylabel("Sales")

plt.show()
```

```python
df["Cost Percentage"] = (df["Cost"] / df["Sales"]) * 100
```

```python
costly_products = (
    df.groupby("Product Name")["Cost Percentage"]
      .mean()
      .sort_values(ascending=False)
)

costly_products.head(10)
```

```python
plt.figure(figsize=(12,6))

costly_products.head(10).plot(kind="bar")

plt.title("Top 10 Highest Cost Percentage Products")
plt.xlabel("Product Name")
plt.ylabel("Average Cost Percentage")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

## Business Insight

- Products with a high Cost Percentage consume a large portion of sales revenue.
- These products should be reviewed for supplier negotiations or manufacturing cost reduction.
- If costs cannot be reduced, pricing strategies should be reconsidered.

```python
low_margin = (
    df.groupby("Product Name")["Gross Margin (%)"]
      .mean()
      .sort_values()
)

low_margin.head(10)
```

```python
plt.figure(figsize=(12,6))

low_margin.head(10).plot(kind="bar", color="red")

plt.title("Top 10 Lowest Margin Products")
plt.xlabel("Product Name")
plt.ylabel("Average Gross Margin (%)")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

## Recommendation

- Products with consistently low gross margins should be reviewed.
- Possible actions:
  - Increase selling price.
  - Negotiate supplier costs.
  - Reduce production costs.
  - Consider discontinuation if profitability remains low.
